## Soal 2. Energi Harian dan Capacity Factor PLTD

```python
# ===== SOAL 2 =====
# a. Simpan nilai dalam variabel
daya_kw = 450            # daya PLTD (kW)
jam = 8                  # lama beroperasi (jam)
energi_aktual = 7200     # energi aktual 24 jam (kWh)
kapasitas_kw = 400       # kapasitas terpasang (kW)

# Hitung energi harian: energi = daya x jam
energi = daya_kw * jam
print(f"Energi harian = {energi} kWh")

# b. Hitung capacity factor (%) dan tampilkan 2 angka desimal
capacity_factor = (energi_aktual / (kapasitas_kw * 24)) * 100
print(f"Capacity factor = {capacity_factor:.2f}%")

# Penjelasan Soal 2:
# - Data awal disimpan dalam variabel supaya mudah dipakai ulang.
# - Energi harian didapat dari daya dikali jam, yaitu 450 x 8 = 3600 kWh.
# - Capacity factor adalah energi aktual dibagi energi maksimum teoritis
#   (kapasitas x 24 jam), lalu dikali 100 untuk menjadi persen.
# - Pada f-string, ":.2f" membulatkan nilai ke 2 angka di belakang koma.
```

## Soal 4. Data Tekanan Uap Turbin

```python
# ===== SOAL 4 =====
tekanan = [12.4, 13.1, 12.8, 14.2, 13.5]

# a. Tekanan pada pembacaan ke-2 (indeks mulai dari 0, jadi indeks 1)
print("Pembacaan ke-2:", tekanan[1])

# b. Dua pembacaan terakhir dengan slicing indeks negatif
print("Dua pembacaan terakhir:", tekanan[-2:])

# c. Tambahkan dua pembacaan baru (13.9 lalu 12.6), tampilkan jumlah data
tekanan.append(13.9)
tekanan.append(12.6)
print("Jumlah data setelah penambahan:", len(tekanan))

# Penjelasan Soal 4:
# - List menyimpan banyak nilai dalam satu variabel.
# - Indeks dimulai dari 0, jadi pembacaan ke-2 diambil dengan tekanan[1].
# - tekanan[-2:] mengambil data dari elemen kedua dari belakang sampai akhir.
# - append() menambah elemen di akhir list, dan di sini dipanggil dua kali
#   berurutan.
# - len() menghitung jumlah elemen, yaitu 5 + 2 = 7.
```

## Soal 6. Status Operasi Boiler

```python
# ===== SOAL 6 =====
suhu = float(input("Masukkan suhu uap (C): "))
tekanan_boiler = float(input("Masukkan tekanan (bar): "))

# Kondisi TRIP diperiksa paling awal karena paling berbahaya
if suhu >= 550 or tekanan_boiler >= 14:
    print("STATUS: TRIP")
elif suhu > 520 and tekanan_boiler > 12:
    print("STATUS: WARNING")
else:
    print("STATUS: NORMAL")

# Penjelasan Soal 6:
# - input() selalu menghasilkan string, jadi hasilnya diubah dengan float().
# - TRIP memakai "or", jadi cukup salah satu syarat terpenuhi (suhu >= 550
#   atau tekanan >= 14).
# - WARNING memakai "and", jadi suhu > 520 dan tekanan > 12 harus terpenuhi
#   sekaligus.
# - TRIP diperiksa lebih dulu karena pada if-elif-else hanya blok pertama
#   yang benar yang dijalankan. Kalau WARNING ditaruh di atas, data seperti
#   suhu 540 dan tekanan 14 akan terbaca WARNING padahal seharusnya TRIP.
# - Contoh uji: (550,10) = TRIP, (540,14) = TRIP, (530,13) = WARNING,
#   (530,11) = NORMAL, (520,13) = NORMAL.
```

## Soal 8. Pengosongan Tangki Bahan Bakar

```python
# ===== SOAL 8 =====
volume = 100

while volume > 0:
    volume -= 20                       # kurangi 20 liter
    print(f"Volume: {volume} liter")   # cetak setelah dikurangi
    if volume == 20:
        print("WARNING: Hampir kosong!")

print("TANGKI KOSONG - Pengosongan berhenti.")

# Penjelasan Soal 8:
# - while mengulang selama volume > 0.
# - "volume -= 20" mengurangi 20 liter tiap putaran:
#   100 -> 80 -> 60 -> 40 -> 20 -> 0.
# - Volume dicetak setelah dikurangi, sesuai perintah soal.
# - if volume == 20 mencetak peringatan tepat saat volume tinggal 20 liter.
# - Ketika volume mencapai 0, kondisi while menjadi salah dan loop berhenti.
#   Pesan tangki kosong dicetak sesudahnya, di luar loop (tanpa indentasi).
```

## Soal 10. Pembacaan Suhu Bearing dengan Sensor Rusak

```python
# ===== SOAL 10 =====
data = ["62.5", "58.0", "81.3", "N/A", "66.0"]
jumlah_valid = 0

for n, teks in enumerate(data, start=1):   # nomor titik mulai dari 1
    try:
        nilai = float(teks)                # ubah teks menjadi float
    except ValueError:
        print(f"Titik {n}: data tidak valid - pemeriksaan dihentikan")
        break                              # hentikan perulangan
    jumlah_valid += 1                      # data berhasil dibaca
    if nilai > 70.0:
        print(f"Warning: Titik {n} suhu bearing tinggi!")

print(f"Jumlah data valid: {jumlah_valid}")

# Penjelasan Soal 10:
# - for dengan enumerate(data, start=1) menelusuri data sambil memberi
#   nomor titik yang dimulai dari 1.
# - try-except ValueError menangkap kegagalan float() saat data bukan
#   angka (misalnya "N/A"), jadi program tidak crash.
# - break menghentikan perulangan begitu ada data tidak valid, sehingga
#   titik 5 tidak diperiksa.
# - jumlah_valid bertambah setiap kali data berhasil diubah menjadi angka.
# - Nilai di atas 70.0 memicu pesan warning, yaitu titik 3 (81.3).
# - Hasil akhir: 3 data valid (62.5, 58.0, 81.3).
```
