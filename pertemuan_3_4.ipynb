```python
#praktik 2

# Operator %d untuk integer
umur = 20
print("Umur saya %d tahun" % (umur))

# Operator %f untuk float
tinggi = 170.5
print("Tinggi badan saya %f cm" % (tinggi))

# Operator %s untuk string
nama = "irwan"
print("Nama saya %s" % (nama))

# Operator %x untuk hexadesimal
angka_hex = 255
print("Nilai hexadesimal dari 255 adalah %x" % (angka_hex))

# Operator %o untuk octal
angka_oct = 64
print("Nilai octal dari 64 adalah %o" % (angka_oct))
```

```python
#praktik 3

total_detik = int(input("Masukkan total detik: "))

jam = total_detik // 3600
sisa_detik = total_detik % 3600
menit = sisa_detik // 60
detik = sisa_detik % 60

print(f"Hasil Konversi: {jam} Jam, {menit} Menit, {detik} Detik")
```

```python
#praktik 4

tugas_1 = float(input("Masukkan nilai tugas 1: "))
tugas_2 = float(input("Masukkan nilai tugas 2: "))
tugas_3 = float(input("Masukkan nilai tugas 3: "))
tugas_4 = float(input("Masukkan nilai tugas 4: "))
tugas_5 = float(input("Masukkan nilai tugas 5: "))

uts = float(input("Masukkan nilai UTS: "))
uas = float(input("Masukkan nilai UAS: "))

rata_tugas = (tugas_1 + tugas_2 + tugas_3 + tugas_4 + tugas_5) / 5
nilai_mahasiswa = (0.3 * rata_tugas) + (0.3 * uts) + (0.4 * uas)

standar_kelulusan = float(input("Masukkan standar kelulusan: "))

apakah_lulus = nilai_mahasiswa >= standar_kelulusan
perlu_remedial = nilai_mahasiswa < standar_kelulusan

print(f"Nilai Mahasiswa: {nilai_mahasiswa}")
print(f"Standar Kelulusan: {standar_kelulusan}")
print(f"Apakah Mahasiswa Lulus? {apakah_lulus}")
print(f"Perlu Remedial? {perlu_remedial}")
```
